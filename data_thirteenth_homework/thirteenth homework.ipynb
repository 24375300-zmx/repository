```python
from typing import List, Dict, Deque
from collections import deque

# 顶点标识与邻接关系
nodes = ["A", "B", "C", "D", "E"]
graph_map = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "E"],
    "D": ["B"],
    "E": ["C"]
}

def depth_first_search(start: str, adj: Dict[str, List[str]]) -> List[str]:
    """深度优先遍历，递归实现"""
    visited_set = set()
    path_record = []

    def dfs_step(cur):
        if cur in visited_set:
            return
        visited_set.add(cur)
        path_record.append(cur)
        for neighbor in adj[cur]:
            dfs_step(neighbor)
    dfs_step(start)
    return path_record

def breadth_first_search(start: str, adj: Dict[str, List[str]]) -> List[str]:
    """广度优先遍历，队列实现"""
    visited_set = set()
    path_record = []
    queue = deque([start])
    visited_set.add(start)

    while queue:
        current = queue.popleft()
        path_record.append(current)
        for neighbor in adj[current]:
            if neighbor not in visited_set:
                visited_set.add(neighbor)
                queue.append(neighbor)
    return path_record

def format_print_path(title: str, seq: List[str]):
    """格式化输出遍历序列"""
    link_str = " → ".join(seq)
    print(f"{title}：{link_str}")

if __name__ == "__main__":
    dfs_result = depth_first_search("A", graph_map)
    bfs_result = breadth_first_search("A", graph_map)

    format_print_path("DFS遍历序列", dfs_result)
    format_print_path("BFS遍历序列", bfs_result)
```
