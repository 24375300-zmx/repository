```python
import random
from typing import List

# 随机基准 单路划分快排
def quick_sort(arr: List[int]) -> List[int]:
    def _sort(low: int, high: int) -> None:
        if high - low <= 1:
            return
        pivot_idx = random.randint(low, high - 1)
        arr[pivot_idx], arr[high - 1] = arr[high - 1], arr[pivot_idx]
        pivot = arr[high - 1]

        i = low - 1
        for j in range(low, high - 1):
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        arr[i + 1], arr[high - 1] = arr[high - 1], arr[i + 1]
        mid = i + 1

        _sort(low, mid)
        _sort(mid + 1, high)

    copy_arr = arr.copy()
    _sort(0, len(copy_arr))
    return copy_arr
```

```python
# 随机基准 三路划分快排
def quick_sort_3way(arr: List[int]) -> List[int]:
    def _sort(low: int, high: int) -> None:
        if high - low <= 1:
            return
        pivot_idx = random.randint(low, high - 1)
        arr[low], arr[pivot_idx] = arr[pivot_idx], arr[low]
        pivot = arr[low]

        lt, gt, i = low, high - 1, low + 1
        while i <= gt:
            if arr[i] < pivot:
                arr[lt], arr[i] = arr[i], arr[lt]
                lt += 1
                i += 1
            elif arr[i] > pivot:
                arr[i], arr[gt] = arr[gt], arr[i]
                gt -= 1
            else:
                i += 1
        _sort(low, lt)
        _sort(gt + 1, high)

    copy_arr = arr.copy()
    _sort(0, len(copy_arr))
    return copy_arr
```

```python
# 统一测试工具函数（修复格式化报错）
def run_test(sort_func, name: str, cases: List[List[int]]):
    print(f"\n========== {name} ==========")
    for raw in cases:
        res = sort_func(raw)
        # 先转字符串再设置宽度
        print(f"原数组: {str(raw):<22} → 排序后: {res}")
```

```python
# 全部测试用例
test_cases = [
    [3, 6, 8, 10, 1, 2, 1],
    [5, 3, 8, 4, 2, 7, 1, 10],
    [1, 2, 3, 4, 5],
    [10, 9, 8, 7, 6],
    [1],
    [],
    [3, 6, 8, 10, 1, 2, 1, 1, 1]
]
```

```python
# 执行测试
if __name__ == "__main__":
    run_test(quick_sort, "单路随机基准快排", test_cases)
    run_test(quick_sort_3way, "三路随机基准快排", test_cases)
```
