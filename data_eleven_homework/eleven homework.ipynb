```python
import heapq
from typing import List

class MedianCalculator:
    def __init__(self):
        # 存储前半段数据，模拟大顶堆（存储负数）
        self.left_heap: List[int] = []
        # 存储后半段数据，原生小顶堆
        self.right_heap: List[int] = []

    def insert_data(self, value: int) -> None:
        # 优先放入左侧堆
        heapq.heappush(self.left_heap, -value)
        # 左侧最大值转移至右侧堆，保证有序划分
        largest_left = -heapq.heappop(self.left_heap)
        heapq.heappush(self.right_heap, largest_left)

        # 平衡两堆长度，左侧堆元素数量等于或比右侧多一个
        if len(self.right_heap) > len(self.left_heap):
            smallest_right = heapq.heappop(self.right_heap)
            heapq.heappush(self.left_heap, -smallest_right)

    def get_middle_value(self) -> float:
        # 总数为奇数，中位数取自左侧堆顶
        if len(self.left_heap) > len(self.right_heap):
            return float(-self.left_heap[0])
        # 总数偶数，取两堆顶平均值
        return (-self.left_heap[0] + self.right_heap[0]) / 2

if __name__ == "__main__":
    calculator = MedianCalculator()
    test_data = [3, 1, 4, 1, 5]
    separator = "-" * 36
    print("数据流中位数计算演示")
    print(separator)

    for idx, num in enumerate(test_data):
        calculator.insert_data(num)
        mid = calculator.get_middle_value()
        print(f"第{idx+1}次插入数字 {num}，当前中位数 = {mid}")
```
