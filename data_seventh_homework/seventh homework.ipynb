```python
from typing import Optional, List
from collections import deque
```

```python
class BiNode:
    __slots__ = ("num", "l_sub", "r_sub")
    def __init__(self, val: int = 0, left: Optional["BiNode"] = None, right: Optional["BiNode"] = None):
        self.num = val
        self.l_sub = left
        self.r_sub = right
```

```python
# 层序列表构建二叉树
def create_tree(data: List[Optional[int]]) -> Optional[BiNode]:
    if not data or data[0] is None:
        return None
    root = BiNode(data[0])
    q = deque([root])
    idx = 1
    total = len(data)
    while q and idx < total:
        cur = q.popleft()
        # 左子节点
        if idx < total and data[idx] is not None:
            cur.l_sub = BiNode(data[idx])
            q.append(cur.l_sub)
        idx += 1
        # 右子节点
        if idx < total and data[idx] is not None:
            cur.r_sub = BiNode(data[idx])
            q.append(cur.r_sub)
        idx += 1
    return root
```

```python
# 计算树深度
def max_depth(node: Optional[BiNode]) -> int:
    if not node:
        return 0
    ld = max_depth(node.l_sub)
    rd = max_depth(node.r_sub)
    return max(ld, rd) + 1

# 紧凑版二叉树绘制，缩小间距，输出简洁
def show_tree(root: Optional[BiNode]) -> None:
    if not root:
        print("当前二叉树为空")
        return
    depth = max_depth(root)
    # 每层基础缩进，控制紧凑程度
    base_gap = 4
    lines = []

    def fill(node: Optional[BiNode], level: int, offset: int):
        if not node:
            return
        # 补充当前层级空白行
        while len(lines) <= level:
            lines.append(" " * offset)
        row = lines[level]
        # 填充节点数字
        if len(row) < offset:
            row += " " * (offset - len(row))
        row = row[:offset] + str(node.num) + row[offset+1:]
        lines[level] = row

        diff = depth - level - 1
        space = base_gap * (2 ** diff) // 2
        if space <= 0:
            return

        link_line = level + 1
        while len(lines) <= link_line:
            lines.append(" " * offset)
        # 左斜杠
        if node.l_sub:
            left_pos = offset - space
            link_str = lines[link_line]
            if len(link_str) < left_pos:
                link_str += " " * (left_pos - len(link_str))
            link_str = link_str[:left_pos] + "╱" + link_str[left_pos+1:]
            lines[link_line] = link_str
            fill(node.l_sub, level + 2, left_pos)
        # 右斜杠
        if node.r_sub:
            right_pos = offset + space
            link_str = lines[link_line]
            if len(link_str) < right_pos:
                link_str += " " * (right_pos - len(link_str))
            link_str = link_str[:right_pos] + "╲" + link_str[right_pos+1:]
            lines[link_line] = link_str
            fill(node.r_sub, level + 2, right_pos)

    start_offset = (2 ** depth) * base_gap // 2
    fill(root, 0, start_offset)

    print("-" * 40)
    print("        二叉树可视化结构图")
    print("-" * 40)
    for line in lines:
        print(line.rstrip())
```

```python
# 测试
if __name__ == "__main__":
    arr = [10, 5, 15, 3, 7, None, 20]
    tree = create_tree(arr)
    show_tree(tree)
```
