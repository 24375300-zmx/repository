```python
from typing import Optional

class TreeNode:
    __slots__ = ("val", "left", "right")
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left: Optional[TreeNode] = left
        self.right: Optional[TreeNode] = right

class BinarySearchTree:
    def __init__(self):
        self.root: Optional[TreeNode] = None

    def insert(self, val: int) -> None:
        """插入数值构建二叉搜索树"""
        if self.root is None:
            self.root = TreeNode(val)
            return
        self._insert_recursion(self.root, val)

    def _insert_recursion(self, node: TreeNode, val: int) -> None:
        if val < node.val:
            if node.left is None:
                node.left = TreeNode(val)
            else:
                self._insert_recursion(node.left, val)
        elif val > node.val:
            if node.right is None:
                node.right = TreeNode(val)
            else:
                self._insert_recursion(node.right, val)

    def remove(self, val: int, replace_type: str = "successor") -> None:
        """删除指定节点，支持前驱/后继两种替换方式"""
        self.root = self._remove_recursion(self.root, val, replace_type)

    def _remove_recursion(self, node: Optional[TreeNode], val: int, replace_type: str) -> Optional[TreeNode]:
        if node is None:
            return None

        if val < node.val:
            node.left = self._remove_recursion(node.left, val, replace_type)
        elif val > node.val:
            node.right = self._remove_recursion(node.right, val, replace_type)
        else:
            # 叶子节点直接删除
            if node.left is None and node.right is None:
                return None
            # 单分支节点直接替代
            if node.left is None:
                return node.right
            if node.right is None:
                return node.left
            # 双分支节点替换逻辑
            if replace_type == "predecessor":
                max_left_node = self.get_max_node(node.left)
                node.val = max_left_node.val
                node.left = self._remove_recursion(node.left, max_left_node.val, replace_type)
            else:
                min_right_node = self.get_min_node(node.right)
                node.val = min_right_node.val
                node.right = self._remove_recursion(node.right, min_right_node.val, replace_type)
        return node

    def get_min_node(self, node: TreeNode) -> TreeNode:
        """获取子树最小值节点"""
        while node.left:
            node = node.left
        return node

    def get_max_node(self, node: TreeNode) -> TreeNode:
        """获取子树最大值节点"""
        while node.right:
            node = node.right
        return node

    def print_tree(self) -> None:
        """分层打印二叉树，优化间距，输出紧凑美观"""
        if self.root is None:
            print("二叉搜索树为空")
            return
        tree_lines = self.generate_draw_lines(self.root)
        for line in tree_lines:
            print(line)

    def generate_draw_lines(self, node: Optional[TreeNode]) -> list[str]:
        if node is None:
            return []
        node_text = str(node.val)
        left_lines = self.generate_draw_lines(node.left)
        right_lines = self.generate_draw_lines(node.right)

        # 叶子节点处理
        if not left_lines and not right_lines:
            block_width = max(4, len(node_text) + 2)
            offset = (block_width - len(node_text)) // 2
            return [" " * offset + node_text + " " * (block_width - offset - len(node_text))]

        # 仅左子树
        if not right_lines:
            base_width = len(left_lines[0])
            pad = (base_width - len(node_text)) // 2
            top_line = " " * pad + node_text + " " * (base_width - pad - len(node_text))
            return [top_line] + left_lines

        # 仅右子树
        if not left_lines:
            base_width = len(right_lines[0])
            pad = (base_width - len(node_text)) // 2
            top_line = " " * pad + node_text + " " * (base_width - pad - len(node_text))
            return [top_line] + right_lines

        # 左右子树同时存在
        left_w = len(left_lines[0])
        right_w = len(right_lines[0])
        space_gap = 3
        total_width = left_w + space_gap + right_w

        # 居中放置当前节点
        mid_pad = (total_width - len(node_text)) // 2
        header_line = " " * mid_pad + node_text + " " * (total_width - mid_pad - len(node_text))

        merge_list = []
        max_h = max(len(left_lines), len(right_lines))
        for i in range(max_h):
            l_part = left_lines[i] if i < len(left_lines) else " " * left_w
            r_part = right_lines[i] if i < len(right_lines) else " " * right_w
            merge_list.append(l_part + " " * space_gap + r_part)
        return [header_line] + merge_list

if __name__ == "__main__":
    data_list = [50, 30, 70, 20, 40, 60, 80]
    separator = "=" * 44

    print(separator)
    print("初始完整二叉搜索树")
    print(separator)
    tree1 = BinarySearchTree()
    for num in data_list:
        tree1.insert(num)
    tree1.print_tree()

    print(f"\n{separator}")
    print("删除根节点50，采用后继替换策略")
    print(separator)
    tree2 = BinarySearchTree()
    for num in data_list:
        tree2.insert(num)
    tree2.remove(50, replace_type="successor")
    tree2.print_tree()

    print(f"\n{separator}")
    print("删除根节点50，采用前驱替换策略")
    print(separator)
    tree3 = BinarySearchTree()
    for num in data_list:
        tree3.insert(num)
    tree3.remove(50, replace_type="predecessor")
    tree3.print_tree()
```
