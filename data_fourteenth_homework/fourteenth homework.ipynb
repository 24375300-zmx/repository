```python
from typing import List, Tuple

# 存储最小生成树边集合
mst_edge_list = [
    ("A", "B"),
    ("A", "D"),
    ("B", "C"),
    ("B", "E"),
    ("E", "F")
]

def print_mst_edges(edge_set: List[Tuple[str, str]]):
    """格式化输出最小生成树所有边"""
    print("MST包含的全部边：")
    for idx, edge in enumerate(edge_set, start=1):
        u, v = edge
        print(f"{idx}. ({u}, {v})")

if __name__ == "__main__":
    print_mst_edges(mst_edge_list)
```
