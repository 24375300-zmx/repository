```python
from typing import Optional, List

class Node:
    __slots__ = ("data", "next")  # 节省内存开销
    def __init__(self, data: int):
        self.data: int = data
        self.next: Optional[Node] = None

class LinkedList:
    def __init__(self):
        self.head: Optional[Node] = None

    def append(self, data: int) -> None:
        """尾部追加节点"""
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node

    def insert_at_head(self, data: int) -> None:
        """头部插入节点"""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    def delete_node(self, key: int) -> bool:
        """删除第一个值等于key的节点，成功返回True，不存在返回False"""
        cur = self.head
        prev = None
        while cur and cur.data != key:
            prev, cur = cur, cur.next
        if not cur:
            return False
        if prev:
            prev.next = cur.next
        else:
            self.head = cur.next
        return True

    def delete_at_position(self, pos: int) -> bool:
        """按索引删除节点，pos为负数/越界返回False"""
        if not self.head or pos < 0:
            return False
        cur = self.head
        if pos == 0:
            self.head = cur.next
            return True
        prev = None
        idx = 0
        while cur and idx != pos:
            prev, cur = cur, cur.next
            idx += 1
        if not cur:
            return False
        prev.next = cur.next
        return True

    def get_length(self) -> int:
        """获取链表长度"""
        length = 0
        cur = self.head
        while cur:
            length += 1
            cur = cur.next
        return length

    def to_list(self) -> List[int]:
        """链表转为普通列表，方便校验"""
        res = []
        cur = self.head
        while cur:
            res.append(cur.data)
            cur = cur.next
        return res

    def clear(self) -> None:
        """清空整个链表"""
        self.head = None

    def print_list(self) -> None:
        """打印链表"""
        print(" ".join(map(str, self.to_list())))

    def __repr__(self) -> str:
        return f"LinkedList{self.to_list()}"
```

```python
# 测试代码
if __name__ == '__main__':
    llist = LinkedList()
    llist.append(1)
    llist.append(2)
    llist.append(3)
    llist.append(4)
    print("原始链表:", llist)

    llist.delete_node(3)
    print("删除值为3的节点后:", llist)

    llist.delete_at_position(1)
    print("删除位置1的节点后:", llist)

    llist.insert_at_head(5)
    print("头部插入5后:", llist)

    llist.delete_at_position(0)
    print("删除头部节点后:", llist)

    # 额外边界测试
    print(f"链表当前长度: {llist.get_length()}")
    print("尝试删除不存在的值9:", llist.delete_node(9))
    print("尝试删除非法位置-1:", llist.delete_at_position(-1))
    llist.clear()
    print("清空链表后:", llist)
```
